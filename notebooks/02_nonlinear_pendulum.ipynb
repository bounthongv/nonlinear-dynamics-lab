# Nonlinear Pendulum — Interactive Exploration

Unlike the simple harmonic oscillator taught in introductory physics, the **real pendulum** is nonlinear:

$$\ddot{\theta} + \gamma \dot{\theta} + \omega_0^2 \sin(\theta) = F \cos(\omega t)$$

where $\omega_0 = \sqrt{g/L}$ is the natural frequency.

### Phenomena to Explore
- **Small vs large angle** — period depends on amplitude
- **Phase portraits** — energy visualized
- **Damping** — energy dissipation
- **Forced oscillations** — route to chaos

```python
import numpy as np
import matplotlib.pyplot as plt
from python.nonlinear_pendulum import simulate, period_estimate, small_angle_period

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. Small-Angle vs Large-Angle Oscillations

The small-angle approximation ($\sin\theta \approx \theta$) gives a period independent of amplitude.
The real pendulum shows **longer periods for larger amplitudes**.

```python
t, th_small, _ = simulate(0.1, 0.0, 10.0)   # ~5.7°
t, th_med,   _ = simulate(1.0, 0.0, 10.0)   # ~57°
t, th_large, _ = simulate(2.0, 0.0, 10.0)   # ~115°

plt.figure(figsize=(12, 5))
plt.plot(t, th_small, label=f'θ₀ = 0.1 rad ({period_estimate(0.1):.3f}s)')
plt.plot(t, th_med,   label=f'θ₀ = 1.0 rad ({period_estimate(1.0):.3f}s)')
plt.plot(t, th_large, label=f'θ₀ = 2.0 rad ({period_estimate(2.0):.3f}s)')
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.3)
plt.xlabel('Time (s)')
plt.ylabel('θ (rad)')
plt.title('Nonlinear Pendulum — Amplitude Affects Period')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print(f'Small-angle period (theory): {small_angle_period():.4f}s')
print('Note: As amplitude grows, period increases — this is impossible')
print('      in the linear approximation.')
```

## 2. Period vs Amplitude

The period increases with amplitude. At θ₀ = π, the period diverges to infinity.

```python
amplitudes = np.linspace(0.05, 2.8, 50)
periods = []
for a in amplitudes:
    p = period_estimate(a, t_max=60.0)
    if p is not None:
        periods.append(p)
    else:
        periods.append(np.nan)

plt.figure(figsize=(10, 6))
plt.plot(amplitudes[:len(periods)], periods, 'b-', linewidth=1.5)
plt.axhline(y=small_angle_period(), color='r', linestyle='--',
            label=f'Small-angle approx: {small_angle_period():.3f}s')
plt.xlabel('Initial amplitude θ₀ (rad)')
plt.ylabel('Period (s)')
plt.title('Pendulum Period vs Amplitude')
plt.xlim(0, 3.0)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 3. Phase Portrait — Undamped

In the $\theta$–$\dot{\theta}$ plane, the undamped pendulum traces closed curves.
Each curve corresponds to a constant energy.

```python
plt.figure(figsize=(8, 7))
thetas0 = [0.5, 1.0, 1.5, 2.0, 2.5, 3.0]
for th0 in thetas0:
    t, th, om = simulate(th0, 0.0, 30.0)
    plt.plot(th, om, label=f'θ₀ = {th0}', linewidth=0.8)
plt.xlabel('θ (rad)')
plt.ylabel('θ̇ (rad/s)')
plt.title('Phase Portrait — Undamped Pendulum')
plt.legend()
plt.grid(alpha=0.3)
plt.axis('equal')
plt.show()
```

## 4. Damped Pendulum

Energy dissipates, and the spiral converges to the stable fixed point at (0, 0).

```python
gammas = [0.1, 0.3, 0.6]
plt.figure(figsize=(12, 4))
for i, g in enumerate(gammas):
    t, th, om = simulate(2.0, 0.0, 30.0, gamma=g)
    plt.subplot(1, 3, i+1)
    plt.plot(th, om, linewidth=0.7)
    plt.xlabel('θ')
    plt.ylabel('θ̇')
    plt.title(f'γ = {g}')
    plt.grid(alpha=0.3)
    plt.axis('equal')
plt.suptitle('Phase Portrait — Damped Pendulum (Spiral to Fixed Point)')
plt.tight_layout()
plt.show()
```

## 5. Forced + Damped Pendulum — Route to Chaos

With periodic forcing, the pendulum can exhibit **chaotic motion**.
This is the driven, damped pendulum — a classic example in nonlinear dynamics.

```python
t, th, om = simulate(1.0, 0.0, 100.0, gamma=0.5, F=1.2, omega=2.0/3.0)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(t[-2000:], th[-2000:], 'b-', linewidth=0.5)
plt.xlabel('Time (s)')
plt.ylabel('θ (rad)')
plt.title('Forced Pendulum — Time Series')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(th[-2000:], om[-2000:], 'r.', markersize=0.5)
plt.xlabel('θ (rad)')
plt.ylabel('θ̇ (rad/s)')
plt.title('Phase Portrait — Strange Attractor?')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Summary

The nonlinear pendulum reveals phenomena invisible in the linear approximation:
- **Amplitude-dependent period**
- **Phase portrait topology** (center vs saddle points)
- **Damped spiral to equilibrium**
- **Chaotic motion under periodic forcing**
