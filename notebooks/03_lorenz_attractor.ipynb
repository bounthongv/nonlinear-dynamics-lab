# Lorenz Attractor — Interactive Exploration

The Lorenz system is one of the most famous examples of **deterministic chaos**.
Discovered by Edward Lorenz in 1963 while modeling atmospheric convection, it showed
that simple deterministic equations can produce unpredictable behavior.

$$\begin{align*}
\frac{dx}{dt} &= \sigma(y - x) \\
\frac{dy}{dt} &= x(\rho - z) - y \\
\frac{dz}{dt} &= xy - \beta z
\end{align*}$$

Classic parameters (chaotic): $\sigma = 10$, $\beta = 8/3$, $\rho = 28$

```python
import numpy as np
import matplotlib.pyplot as plt
from python.lorenz_attractor import simulate

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. The Butterfly Attractor

The trajectory orbits around two fixed points, forming the iconic butterfly shape.
The system never repeats — it's deterministic but unpredictable.

```python
t, x, y, z = simulate(1.0, 1.0, 1.0, 40.0)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, 'b-', linewidth=0.5, alpha=0.8)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Lorenz Attractor — "The Butterfly"')
plt.tight_layout()
plt.show()
```

## 2. Time Series — Unpredictable Switching

The $x(t)$ and $z(t)$ signals show **irregular switching** between the two lobes of the attractor.
The number of oscillations on each lobe is unpredictable.

```python
plt.figure(figsize=(14, 8))

plt.subplot(3, 1, 1)
plt.plot(t, x, 'b-', linewidth=0.5)
plt.ylabel('x(t)')
plt.title('Lorenz System — Time Series')
plt.grid(alpha=0.3)
plt.xlim(0, 40)

plt.subplot(3, 1, 2)
plt.plot(t, y, 'g-', linewidth=0.5)
plt.ylabel('y(t)')
plt.grid(alpha=0.3)
plt.xlim(0, 40)

plt.subplot(3, 1, 3)
plt.plot(t, z, 'r-', linewidth=0.5)
plt.ylabel('z(t)')
plt.xlabel('Time')
plt.grid(alpha=0.3)
plt.xlim(0, 40)

plt.tight_layout()
plt.show()
```

## 3. 2D Projections

Each projection shows a different face of the attractor.

```python
plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1)
plt.plot(x, y, 'b-', linewidth=0.4)
plt.xlabel('x')
plt.ylabel('y')
plt.title('X-Y Plane')
plt.grid(alpha=0.3)

plt.subplot(1, 3, 2)
plt.plot(x, z, 'r-', linewidth=0.4)
plt.xlabel('x')
plt.ylabel('z')
plt.title('X-Z Plane')
plt.grid(alpha=0.3)

plt.subplot(1, 3, 3)
plt.plot(y, z, 'g-', linewidth=0.4)
plt.xlabel('y')
plt.ylabel('z')
plt.title('Y-Z Plane')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 4. Sensitivity to Initial Conditions

Two trajectories starting almost identically diverge rapidly.
This is the **butterfly effect** — a tiny change in initial conditions
leads to completely different outcomes.

```python
_, x1, y1, z1 = simulate(1.000, 1.000, 1.000, 20.0)
_, x2, y2, z2 = simulate(1.001, 1.000, 1.000, 20.0)  # 0.1% difference in x₀

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(x1, z1, 'b-', linewidth=0.5, label='x₀ = 1.000')
plt.plot(x2, z2, 'r-', linewidth=0.5, label='x₀ = 1.001')
plt.xlabel('x')
plt.ylabel('z')
plt.title('X-Z Projection — Two Nearby Trajectories')
plt.legend(fontsize=8)
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
diff = np.sqrt((x1 - x2)**2 + (y1 - y2)**2 + (z1 - z2)**2)
plt.semilogy(t, diff, 'k-', linewidth=0.8)
plt.xlabel('Time')
plt.ylabel('Distance between trajectories')
plt.title('Exponential Divergence')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
print('The distance grows exponentially — this is the hallmark of chaos.')
```

## 5. Different Parameter Regimes

The behavior of the Lorenz system depends critically on $\rho$:

```python
rhos = [14, 24, 28, 40]
labels = ['ρ=14 (Stable fixed point)', 'ρ=24 (Pre-chaos)',
          'ρ=28 (Chaos — butterfly)', 'ρ=40 (Strong chaos)']

fig, axes = plt.subplots(2, 2, figsize=(12, 10), subplot_kw={'projection': '3d'})
for ax, r, label in zip(axes.flat, rhos, labels):
    _, x, y, z = simulate(1.0, 1.0, 1.0, 30.0, rho=r)
    ax.plot(x, y, z, linewidth=0.5)
    ax.set_title(label, fontsize=10)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

plt.suptitle('Lorenz System — Different ρ Regimes', fontsize=14)
plt.tight_layout()
plt.show()
```

## Summary

The Lorenz attractor demonstrates:
- **Deterministic chaos** — simple equations, unpredictable
- **Strange attractor** — bounded but never repeating
- **Butterfly effect** — sensitive dependence on initial conditions
- **Parameter sensitivity** — chaos emerges above ρ ≈ 24.7
