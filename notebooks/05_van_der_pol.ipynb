# Van der Pol Oscillator — Interactive Exploration

The Van der Pol oscillator is a **self-excited oscillator** with nonlinear damping:

$$\ddot{x} - \mu(1 - x^2)\dot{x} + x = F\cos(\omega t)$$

### Key phenomena
- **Self-excited oscillations** — the system spontaneously oscillates without periodic forcing
- **Limit cycle** — all trajectories converge to the same periodic orbit
- **Relaxation oscillations** — for large $\mu$, slow-fast dynamics with sharp transitions
- **Forced chaos** — with external periodic forcing

First studied by Balthasar van der Pol in the 1920s while modeling vacuum tube circuits.

```python
import numpy as np
import matplotlib.pyplot as plt
from python.van_der_pol import simulate, limit_cycle_amplitude

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. The Limit Cycle

No matter where you start, the system converges to the same **limit cycle** —
a stable periodic orbit. This is self-excited oscillation.

```python
# Two different initial conditions converge to the same limit cycle
_, x1, v1 = simulate(0.1, 0.1, 50.0, mu=1.0)
_, x2, v2 = simulate(3.0, 0.0, 50.0, mu=1.0)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(x1, v1, 'b-', linewidth=0.6, label='x₀=0.1')
plt.plot(x2, v2, 'r-', linewidth=0.6, label='x₀=3.0')
plt.xlabel('x'); plt.ylabel('v')
plt.title('Phase Portrait — Converging to Same Limit Cycle')
plt.legend(); plt.grid(alpha=0.3)
plt.axis('equal')

plt.subplot(1, 2, 2)
plt.plot(x1[-2000:], v1[-2000:], 'b-', linewidth=0.8)
plt.xlabel('x'); plt.ylabel('v')
plt.title('Steady State — Limit Cycle')
plt.grid(alpha=0.3)
plt.axis('equal')
plt.tight_layout()
plt.show()
print('Both trajectories converge to the same limit cycle.')
print('This is a hallmark of self-excited oscillations.')
```

## 2. The Birth of the Limit Cycle

For $\mu < 0$, the oscillator is damped (spiral to rest).
For $\mu > 0$, a limit cycle emerges — this is a **Hopf bifurcation**.

```python
mus = [-0.5, 0.0, 0.5, 2.0]
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, m in zip(axes, mus):
    _, x, v = simulate(0.5, 0.0, 80.0, mu=m)
    ax.plot(x, v, linewidth=0.5)
    ax.set_title(f'μ = {m}')
    ax.set_xlabel('x'); ax.set_ylabel('v')
    ax.grid(alpha=0.3); ax.axis('equal')

plt.suptitle('Hopf Bifurcation — Limit Cycle Emerges at μ = 0', fontsize=14)
plt.tight_layout()
plt.show()
```

## 3. Time Series — Different μ Regimes

At small μ, the waveform is nearly sinusoidal. At large μ,
it becomes a **relaxation oscillation** — slow build-up followed by rapid discharge.

```python
mus = [0.1, 1.0, 3.0, 8.0]
plt.figure(figsize=(14, 10))

for i, m in enumerate(mus):
    t, x, _ = simulate(0.1, 0.1, 40.0, mu=m)
    plt.subplot(4, 1, i+1)
    plt.plot(t, x, linewidth=0.6)
    plt.ylabel('x(t)')
    plt.title(f'μ = {m}')
    plt.grid(alpha=0.3)
    plt.xlim(0, 40)

plt.xlabel('Time')
plt.suptitle('Van der Pol — From Sine Waves to Relaxation Oscillations', fontsize=14)
plt.tight_layout()
plt.show()
print('μ=0.1: Nearly sinusoidal')
print('μ=1.0: Mild distortion')
print('μ=3.0: Relaxation oscillation begins')
print('μ=8.0: Classic relaxation oscillation — sharp spikes')
```

## 4. Forced Van der Pol — Chaos

With periodic forcing, the Van der Pol oscillator can become chaotic —
another classic example from nonlinear dynamics.

```python
t, x, v = simulate(0.1, 0.1, 200.0, mu=1.0, F=0.8, omega=1.2)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.plot(t[-3000:], x[-3000:], 'b-', linewidth=0.5)
plt.xlabel('Time'); plt.ylabel('x')
plt.title('Forced Van der Pol — Time Series')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(x[-5000:], v[-5000:], 'r-', linewidth=0.3)
plt.xlabel('x'); plt.ylabel('v')
plt.title('Phase Portrait')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
print('With strong forcing, the limit cycle becomes a torus,')
print('which can break down into chaos.')
```

## 5. Relaxation Oscillation Waveform

At large μ, the waveform has a distinctive shape: slow accumulation,
then rapid discharge. This is called a **relaxation oscillation**.

```python
t, x, v = simulate(0.1, 0.1, 40.0, mu=5.0)

plt.figure(figsize=(14, 4))
plt.plot(t, x, 'b-', linewidth=0.8)
plt.xlabel('Time')
plt.ylabel('x(t)')
plt.title('Relaxation Oscillation — Van der Pol with μ = 5.0')
plt.grid(alpha=0.3)
plt.xlim(0, 40)
plt.show()
print('Notice: slow drift upward, then rapid spike downward.')
print('This is the signature of relaxation oscillations.')
```

## Summary

The Van der Pol oscillator demonstrates:
- **Self-excited oscillations** — no external forcing needed
- **Hopf bifurcation** — limit cycle emerges at μ = 0
- **Relaxation oscillations** — slow-fast dynamics at large μ
- **Forced chaos** — with periodic driving

This completes the classic set of nonlinear dynamics simulations.
