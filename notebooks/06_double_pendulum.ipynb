# Double Pendulum — Interactive Exploration

The **double pendulum** is one of the simplest physical systems that exhibits
deterministic chaos. Two pendulums connected end-to-end produce motion that
is exquisitely sensitive to initial conditions.

### Key phenomena
- **Deterministic chaos** — tiny changes in initial angle → vastly different trajectories
- **Beautiful patterns** — the trajectory of the second bob is mesmerizing
- **Energy conservation** — total energy is conserved (undamped case)

```python
import numpy as np
import matplotlib.pyplot as plt
from python.double_pendulum import simulate, energy

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. Trajectory of the Double Pendulum

The path traced by the second bob is stunningly complex — yet purely deterministic.

```python
t, th1, th2, w1, w2, x1, y1, x2, y2 = simulate(np.pi, 0.0, 0, 0, 30.0)

plt.figure(figsize=(8, 8))
plt.plot(x2, y2, 'b-', linewidth=0.3, alpha=0.7)
plt.plot(x1, y1, 'r-', linewidth=0.5, alpha=0.5, label='Bob 1')
plt.scatter([0], [0], color='k', s=50, zorder=5, label='Pivot')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Double Pendulum Trajectory — 30 Seconds')
plt.axis('equal')
plt.legend(fontsize=8)
plt.grid(alpha=0.3)
plt.show()
```

## 2. Time Series of Both Angles

The angular motion appears irregular and never quite repeats.

```python
plt.figure(figsize=(14, 6))
plt.subplot(2, 1, 1)
plt.plot(t, th1, 'b-', linewidth=0.5)
plt.ylabel('θ₁ (rad)'); plt.title('Angle of First Pendulum'); plt.grid(alpha=0.3)

plt.subplot(2, 1, 2)
plt.plot(t, th2, 'r-', linewidth=0.5)
plt.ylabel('θ₂ (rad)'); plt.xlabel('Time (s)')
plt.title('Angle of Second Pendulum'); plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Energy Conservation

Total energy should remain constant. Any drift indicates numerical error.

```python
KE, PE, E_total = energy(th1, th2, w1, w2)

plt.figure(figsize=(12, 4))
plt.plot(t, KE, 'g-', linewidth=0.6, label='Kinetic')
plt.plot(t, PE, 'b-', linewidth=0.6, label='Potential')
plt.plot(t, E_total, 'k--', linewidth=1, label='Total')
plt.xlabel('Time (s)')
plt.ylabel('Energy')
plt.title('Energy Conservation')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

drift = abs(E_total[-1] - E_total[0]) / max(abs(E_total[0]), 1e-10) * 100
print(f'Energy drift: {drift:.6f}%' if abs(E_total[0]) > 1e-10 else 'Energy drift: negligible (E₀≈0)')
```

## 4. Initial Condition Sensitivity

Two runs starting 0.001 radians apart diverge rapidly — the hallmark of chaos.

```python
_, _, _, _, _, _, _, x2a, y2a = simulate(np.pi, 0.0, 0, 0, 20.0)
_, _, _, _, _, _, _, x2b, y2b = simulate(np.pi + 0.001, 0.0, 0, 0, 20.0)

diff = np.sqrt((x2a - x2b)**2 + (y2a - y2b)**2)

plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.semilogy(t, diff, 'k-', linewidth=0.8)
plt.xlabel('Time (s)')
plt.ylabel('Distance between trajectories')
plt.title('Exponential Divergence')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(x2a, y2a, 'b-', lw=0.4, alpha=0.7, label='θ₁₀ = π')
plt.plot(x2b, y2b, 'r-', lw=0.4, alpha=0.7, label='θ₁₀ = π + 0.001')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Two Nearly Identical Starts Diverge')
plt.axis('equal'); plt.legend(fontsize=8); plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
print(f'Final separation: {diff[-1]:.4f} (from initial 0.001)')
```

## 5. Animated Snapshot

Plot the pendulum position at several moments to see the structure.

```python
plt.figure(figsize=(10, 10))
for i in range(0, len(t), 200):
    plt.plot([0, x1[i]], [0, y1[i]], 'r-', linewidth=0.5, alpha=0.3)
    plt.plot([x1[i], x2[i]], [y1[i], y2[i]], 'b-', linewidth=0.5, alpha=0.3)
    plt.scatter(x1[i], y1[i], c='r', s=10, alpha=0.3)
    plt.scatter(x2[i], y2[i], c='b', s=10, alpha=0.3)
plt.scatter([0], [0], color='k', s=50, zorder=5)
plt.xlabel('x'); plt.ylabel('y')
plt.title('Double Pendulum — Snapshot Every 0.2s')
plt.axis('equal')
plt.grid(alpha=0.3)
plt.show()
```

## Summary

The double pendulum demonstrates the essence of nonlinear dynamics:
- **Simple rules, complex behavior** — only 2 masses, 2 arms, yet motion is unpredictable
- **Butterfly effect** — 0.001 rad difference becomes meters of divergence
- **Energy conservation** — checkable even in a chaotic system
- **Deterministic chaos** — the equations are known, the outcome is not

This is a classic example of nonlinear dynamics — the kind explored in
my 1995 book *Ordnung und Chaos bei nichtlinearen Schwingungen*.
