# Duffing Oscillator — Interactive Exploration

The Duffing oscillator is a **forced nonlinear oscillator** that exhibits
a rich variety of dynamical behavior:

$$\ddot{x} + \delta \dot{x} + \alpha x + \beta x^3 = \gamma \cos(\omega t)$$

### What makes it special
- **Double-well potential** (for $\alpha < 0, \beta > 0$) — the particle can oscillate in one well or jump between wells
- **Period-doubling route to chaos** — as forcing amplitude increases
- **Amplitude jump resonance** — the response amplitude can jump suddenly
- **Coexisting attractors** — the same parameters can produce different behaviors depending on initial conditions

```python
import numpy as np
import matplotlib.pyplot as plt
from python.duffing_oscillator import simulate, potential, frequency_response

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. The Double-Well Potential

With $\alpha < 0$ and $\beta > 0$, the Duffing oscillator has two stable equilibrium points
(the two wells) separated by an unstable equilibrium (the barrier).

```python
x_plot = np.linspace(-2.5, 2.5, 200)
V = potential(x_plot)

plt.figure(figsize=(10, 5))
plt.plot(x_plot, V, 'b-', linewidth=2)
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.3)
plt.axvline(x=-1, color='r', linestyle=':', alpha=0.5, label='Left well')
plt.axvline(x=1, color='g', linestyle=':', alpha=0.5, label='Right well')
plt.axvline(x=0, color='orange', linestyle=':', alpha=0.5, label='Barrier')
plt.xlabel('Displacement x')
plt.ylabel('Potential V(x)')
plt.title('Duffing Double-Well Potential (α = -1, β = 1)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print('The particle oscillates in one well (intrawell) or jumps')
print('between wells (interwell) depending on forcing amplitude.')
```

## 2. Intrawell vs Cross-Well Oscillation

At low forcing ($\gamma = 0.3$), the motion is trapped in one well.
At higher forcing ($\gamma = 0.5$), the particle jumps between wells.

```python
_, x1, v1 = simulate(0.5, 0.0, 80.0, gamma=0.3)
_, x2, v2 = simulate(0.5, 0.0, 80.0, gamma=0.5)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x1, v1, 'b-', linewidth=0.4)
plt.xlabel('x'); plt.ylabel('v')
plt.title('Phase Portrait — γ = 0.3 (Intrawell)')
plt.grid(alpha=0.3)

plt.subplot(2, 2, 2)
plt.plot(x2, v2, 'r-', linewidth=0.4)
plt.xlabel('x'); plt.ylabel('v')
plt.title('Phase Portrait — γ = 0.5 (Cross-well)')
plt.grid(alpha=0.3)

plt.subplot(2, 2, 3)
plt.plot(x2[:2000], linewidth=0.6)
plt.xlabel('Time step'); plt.ylabel('x')
plt.title('Time Series — γ = 0.5')
plt.grid(alpha=0.3)

plt.subplot(2, 2, 4)
plt.plot(x1[:2000], linewidth=0.6)
plt.xlabel('Time step'); plt.ylabel('x')
plt.title('Time Series — γ = 0.3')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3. Period-Doubling Route to Chaos

Increase the forcing amplitude and watch the motion go from periodic → period-2 → chaos.

```python
gammas = [0.2, 0.3, 0.4, 0.5]
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, g in zip(axes.flat, gammas):
    _, x, v = simulate(0.5, 0.0, 100.0, gamma=g)
    # Steady state (last half)
    half = len(x) // 2
    ax.plot(x[half:], v[half:], linewidth=0.3)
    ax.set_title(f'γ = {g}')
    ax.set_xlabel('x')
    ax.set_ylabel('v')
    ax.grid(alpha=0.3)

plt.suptitle('Duffing Oscillator — Period Doubling Route to Chaos', fontsize=14)
plt.tight_layout()
plt.show()
print('γ=0.2: Period 1 (simple cycle)')
print('γ=0.3: Period 2 (doubled)')
print('γ=0.4: More complex')
print('γ=0.5: Chaotic (strange attractor)')
```

## 4. Poincaré Section

Strobing the trajectory at the forcing frequency reveals the structure
of the attractor. A single point = period-1. Multiple points = higher period.
A fractal set = chaos.

```python
gamma = 0.5
omega = 1.2
T = 2 * np.pi / omega
dt = 0.01

t, x, v = simulate(0.5, 0.0, 500.0, dt, gamma=gamma, omega=omega)

# Sample at forcing period (Poincaré section)
n_per_period = int(T / dt)
# Discard transient (first 200 periods)
start = 200 * n_per_period
poincare_x = x[start::n_per_period]
poincare_v = v[start::n_per_period]

plt.figure(figsize=(8, 7))
plt.scatter(poincare_x, poincare_v, s=10, c='b', alpha=0.6)
plt.xlabel('x')
plt.ylabel('v')
plt.title(f'Poincaré Section — γ = {gamma} (Chaotic Regime)')
plt.grid(alpha=0.3)
plt.show()
print(f'Collected {len(poincare_x)} Poincaré points')
print('A fractal-like structure indicates a strange attractor.')
```

## 5. Coexisting Attractors

The Duffing oscillator can have **two different steady-state behaviors**
for the same parameters, depending on initial conditions.

```python
# Same parameters, different initial conditions
_, x1, v1 = simulate(0.5, 0.0, 100.0, gamma=0.35)
_, x2, v2 = simulate(-0.5, 0.0, 100.0, gamma=0.35)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(x1[-3000:], v1[-3000:], 'b-', linewidth=0.5)
plt.xlabel('x'); plt.ylabel('v')
plt.title('x₀ = 0.5 (Right well attractor)')
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(x2[-3000:], v2[-3000:], 'r-', linewidth=0.5)
plt.xlabel('x'); plt.ylabel('v')
plt.title('x₀ = -0.5 (Left well attractor)')
plt.grid(alpha=0.3)

plt.suptitle('Coexisting Attractors — Same γ=0.35, Different Initial Conditions')
plt.tight_layout()
plt.show()
print('Bistability: two stable solutions exist simultaneously.')
```

## Summary

The Duffing oscillator reveals:
- **Double-well dynamics** — two stable equilibria
- **Period-doubling route to chaos** — as forcing increases
- **Strange attractors** — in Poincaré section
- **Coexisting attractors** — bistability, hysteresis
