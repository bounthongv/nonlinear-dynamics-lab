# Logistic Map — Interactive Exploration

The logistic map is a deceptively simple equation that produces astonishing complexity:

$$x_{n+1} = r \cdot x_n \cdot (1 - x_n)$$

Despite having only one variable $x$ and one parameter $r$, it exhibits:
- **Fixed points** (stable equilibrium)
- **Period doubling** (bifurcations)
- **Deterministic chaos** (sensitive dependence on initial conditions)
- **Universal scaling** (Feigenbaum constants)

This was one of the core simulations in my 1995 book *Ordnung und Chaos bei nichtlinearen Schwingungen*. Now rebuilt in Python.

```python
import numpy as np
import matplotlib.pyplot as plt
from python.logistic_map import iterate, bifurcation_points, lyapunov_exponent

%matplotlib inline
plt.rcParams['figure.dpi'] = 120
```

## 1. Time Series — See the Behavior Change

Run the cell below and change `r` to see different regimes:

```python
r = 3.9        # Try: 2.5, 3.0, 3.2, 3.45, 3.5, 3.56, 3.9
x0 = 0.5
n = 100

xs = iterate(x0, r, n)

plt.figure(figsize=(12, 4))
plt.plot(range(len(xs)), xs, 'b-', linewidth=0.8)
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.3)
plt.axhline(y=1, color='gray', linestyle='--', alpha=0.3)
plt.xlabel('Iteration n')
plt.ylabel('x_n')
plt.title(f'Logistic Map Time Series — r = {r}')
plt.ylim(-0.05, 1.05)
plt.grid(alpha=0.3)
plt.show()
print(f"Final value after {n} iterations: x_{n} = {xs[-1]:.6f}")
```

**Try these r values and observe:**
- $r = 2.5$ → Fixed point (converges to one value)
- $r = 3.0$ → Approaching bifurcation
- $r = 3.2$ → Period 2 (oscillates between two values)
- $r = 3.45$ → Period 4
- $r = 3.56$ → Period 8 (barely visible)
- $r = 3.9$ → Chaos (never repeats)

## 2. Bifurcation Diagram — The Full Picture

This diagram shows the steady-state behavior for all $r$ at once. It is the signature image of chaos theory.

```python
r_vals = np.linspace(2.5, 4.0, 500)
r_plot, x_plot = bifurcation_points(r_vals, n_transient=1000, n_samples=100)

plt.figure(figsize=(14, 7))
plt.scatter(r_plot, x_plot, s=0.1, c='k', alpha=0.5)
plt.xlabel('Control parameter r')
plt.ylabel('Steady-state x')
plt.title('Bifurcation Diagram of the Logistic Map')
plt.xlim(2.5, 4.0)
plt.ylim(-0.05, 1.05)
plt.grid(alpha=0.3)
plt.show()
```

Notice:
- **Left of 3.0:** Single fixed point
- **3.0–3.45:** Period 2 (first bifurcation)
- **3.45–3.54:** Period 4 (second bifurcation)
- **3.54–3.57:** Period 8, 16, 32... (Feigenbaum point)
- **Beyond 3.57:** Chaos with occasional periodic windows

## 3. Lyapunov Exponent — Measuring Chaos

A positive Lyapunov exponent (λ > 0) indicates chaos — nearby trajectories diverge exponentially.

```python
r_vals = np.linspace(2.5, 4.0, 400)
lyap = [lyapunov_exponent(r) for r in r_vals]

plt.figure(figsize=(12, 5))
plt.plot(r_vals, lyap, 'r-', linewidth=0.8)
plt.axhline(y=0, color='gray', linestyle='--', linewidth=0.5)
plt.fill_between(r_vals, lyap, 0, where=np.array(lyap) > 0,
                 color='red', alpha=0.15, label='Chaos (λ > 0)')
plt.fill_between(r_vals, lyap, 0, where=np.array(lyap) <= 0,
                 color='blue', alpha=0.1, label='Order (λ ≤ 0)')
plt.xlabel('Control parameter r')
plt.ylabel('Lyapunov exponent λ')
plt.title('Lyapunov Exponent of the Logistic Map')
plt.xlim(2.5, 4.0)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## 4. Sensitive Dependence — The Hallmark of Chaos

Two nearly identical starting values diverge rapidly. This is the **butterfly effect**.

```python
r = 3.9
x0_1 = 0.5000
x0_2 = 0.5001  # Just 0.01% difference

xs1 = iterate(x0_1, r, 50)
xs2 = iterate(x0_2, r, 50)

plt.figure(figsize=(12, 5))
plt.plot(range(len(xs1)), xs1, 'b-', label=f'x₀ = {x0_1}', linewidth=0.8)
plt.plot(range(len(xs2)), xs2, 'r--', label=f'x₀ = {x0_2}', linewidth=0.8)
plt.xlabel('Iteration n')
plt.ylabel('x_n')
plt.title(f'Sensitive Dependence on Initial Conditions (r = {r})')
plt.ylim(-0.05, 1.05)
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print(f"After 50 iterations:")
print(f"  x₀ = {x0_1} → {xs1[-1]:.6f}")
print(f"  x₀ = {x0_2} → {xs2[-1]:.6f}")
print(f"  Difference: {abs(xs1[-1] - xs2[-1]):.6f}")
```

## Summary

The logistic map demonstrates that **simple deterministic rules can produce unpredictable behavior** — a core insight of nonlinear dynamics.
